```python
# exercise 16.2

# Convert stationarized TPI Y -> non-stationary constant local-currency Y (compact, robust)
import requests, pickle
from io import BytesIO
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os

# URLs (OG-Core test data)
URL_TPI    = "https://github.com/PSLmodels/OG-Core/blob/master/tests/test_io_data/TPI_vars_baseline.pkl?raw=true"
URL_SS     = "https://github.com/PSLmodels/OG-Core/blob/master/tests/test_io_data/SS_vars_baseline.pkl?raw=true"
URL_PARAMS = "https://github.com/PSLmodels/OG-Core/blob/master/tests/test_io_data/model_params_baseline.pkl?raw=true"

def load_pickle_url(url):
    r = requests.get(url); r.raise_for_status()
    return pickle.loads(BytesIO(r.content).getvalue())

# Load
tpi = load_pickle_url(URL_TPI)
ss  = load_pickle_url(URL_SS)
params = load_pickle_url(URL_PARAMS)

# Stationarized GDP from TPI
Y_station = np.asarray(tpi["Y"], dtype=float).squeeze()
T = Y_station.shape[0]

# factor_ss (try common keys)
factor_ss = None
for k in ("factor_ss", "factor", "factorss"):
    if k in ss:
        factor_ss = float(np.asarray(ss[k]).mean())
        break
if factor_ss is None:
    # fallback: search keys that include 'factor'
    for k in ss:
        if "factor" in k.lower():
            factor_ss = float(np.asarray(ss[k]).mean()); break
if factor_ss is None:
    raise KeyError("factor_ss not found in SS vars. Keys: " + ", ".join(ss.keys()))

# g_n and g_y extraction (handle params being object-like or dict-like)
def get_param(obj, name):
    # try attribute
    if hasattr(obj, name):
        return getattr(obj, name)
    # try dictionary-like
    if isinstance(obj, dict) and name in obj:
        return obj[name]
    # paramtools objects may allow indexing
    try:
        return obj[name]
    except Exception:
        pass
    raise KeyError(f"Parameter '{name}' not found in params.")

g_y_raw = get_param(params, "g_y")
g_n_raw = get_param(params, "g_n")

# normalize g_y to scalar (most tests are scalar)
g_y = float(np.asarray(g_y_raw).squeeze())

# g_n may be scalar or series; produce series length T
g_n_arr = np.asarray(g_n_raw).reshape(-1)
if g_n_arr.size == 0:
    raise ValueError("g_n appears empty.")
if g_n_arr.size == 1:
    g_n = np.repeat(float(g_n_arr[0]), T)
else:
    if g_n_arr.size < T:
        # extend by repeating last element if shorter than T
        g_n = np.concatenate([g_n_arr, np.repeat(g_n_arr[-1], T - g_n_arr.size)])[:T]
    else:
        g_n = g_n_arr[:T]

# Build cumulative growth factors
cum_pop = np.cumprod(1.0 + g_n)         # cumulative product of (1+g_n)
trend_y = np.exp(g_y * np.arange(T))   # exp(g_y * t)

# Convert stationarized Y to non-stationary (constant local currency)
Y_nonstation = Y_station * cum_pop * trend_y * factor_ss

# Save results to CSV
out_csv = "TPI_Y_nonstation_baseline.csv"
df = pd.DataFrame({
    "t": np.arange(T),
    "year": 2025 + np.arange(T),    # adjust start year
    "Y_station": Y_station,
    "cum_pop": cum_pop,
    "trend_y": trend_y,
    "factor_ss": factor_ss,
    "Y_nonstation": Y_nonstation
})
df.to_csv(out_csv, index=False)
print("Saved CSV:", os.path.abspath(out_csv))

# Plot (in billions for readability)
plt.figure(figsize=(10,4))
plt.plot(df["year"], df["Y_nonstation"]/1e9, label="Y (constant local currency, billions)")
plt.plot(df["year"], df["Y_station"], label="Y (stationarized, model units)", alpha=0.6)
plt.xlabel("Year")
plt.ylabel("GDP")
plt.title("Baseline GDP: stationarized vs non-stationary (constant local currency)")
plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

After converting OG-Core’s output into local currenccy non stationary units (including pop growth - productivity growth, and applying steady state scaling factor) the graph shows GDP rises over time in real units, whereas the stationarized series remains flat because trend has been removed in the model.

```python
#Exercise 17.10 — AR(1) regression on annual GDP growth

# --- Load quarterly real GDP for South Africa ---
import datetime as dt
import matplotlib.pyplot as plt
from pandas_datareader import data as web
import pandas as pd
import numpy as np
import statsmodels.api as sm

series_id = "NGDPRNSAXDCZAQ"
start = dt.datetime(1993, 1, 1)
end   = dt.datetime.today()

gdp_q = web.DataReader(series_id, "fred", start, end)

# --- Collapse to annual GDP ---
gdp_annual = gdp_q.groupby(gdp_q.index.year).mean()
gdp_annual.columns = ["GDP"]

# --- Compute annual GDP growth ---
gdp_growth = gdp_annual.pct_change().dropna()
gdp_growth["growth_lag"] = gdp_growth["GDP"].shift(1).pct_change()  # OR simpler:
gdp_growth["growth_lag"] = gdp_growth["GDP"].pct_change().shift(1)

# Clean up
gdp_growth = gdp_growth.dropna()

# Response and regressors
y = gdp_growth["GDP"].pct_change().dropna()
X = gdp_growth["GDP"].pct_change().shift(1).dropna()

# Align indices
df = pd.DataFrame({"y": y, "lag": X}).dropna()

# --- Run AR(1) regression ---
X_ols = sm.add_constant(df["lag"])
model = sm.OLS(df["y"], X_ols).fit()
print(model.summary())

rho = model.params["lag"]
print("\nEstimated AR(1) coefficient (rho):", rho)
```

Exercise 17.10 our rho coefficient on GDP is very small indicating that economic shocks fade relatively quickly over time.  This makes sense given the quick rebound in GDP from our two largest negative shocks in 2007 and 2021

```python
# Exercise 17.12 — scale OG-USA earnings to match a target Gini coefficient
import os
import sys
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize_scalar
from ogusa import income


# -------------------------
# Settings / assumptions
# -------------------------
S = 80        # ages 20..99
J = 7         # ability groups
age_wgts = np.ones(S) / S
abil_wgts = np.ones(J) / J
target_gini = 0.63   # South Africa ~0.63

# -------------------------
# Load earnings matrix 'e' from OG-USA
# -------------------------
e_out = income.get_e_orig(age_wgts, abil_wgts)
e = e_out[0] if isinstance(e_out, tuple) else e_out
e = np.asarray(e)
# truncate if OG-USA returns larger shapes
e = e[:S, :J]

# -------------------------
# helper: compute weighted Gini
#   inputs:
#     vals: 1D array of values (e.g., lifetime incomes by group)
#     w: 1D array of population weights summing to 1
# -------------------------
def weighted_gini(vals, w):
    # require numpy arrays
    vals = np.asarray(vals).astype(float)
    w = np.asarray(w).astype(float)
    if np.any(vals < 0):
        # shift if negatives (shouldn't happen here)
        vals = vals - vals.min() + 1e-12
    # sort by vals
    order = np.argsort(vals)
    vals_s = vals[order]
    w_s = w[order]
    # normalize weights to sum 1
    w_s = w_s / w_s.sum()
    cumw = np.cumsum(w_s)
    # income shares
    total_income = np.sum(vals_s * w_s)
    if total_income <= 0:
        return 0.0
    inc_shares = (vals_s * w_s) / total_income
    cuminc = np.cumsum(inc_shares)
    cuminc_prev = np.concatenate(([0.0], cuminc[:-1]))
    # Weighted Gini: 1 - sum_i w_i * (cuminc_prev + cuminc_i)
    G = 1.0 - np.sum(w_s * (cuminc_prev + cuminc))
    return float(G)

# -------------------------
# helper: compute lifetime incomes by ability group given e matrix
# -------------------------
def lifetime_incomes_from_e(e_mat, age_wgts):
    # e_mat: S x J
    # lifetime income per group j = sum_s age_wgts[s] * e[s,j]
    lifetimes = np.dot(age_wgts, e_mat)  # shape (J,)
    return lifetimes

# -------------------------
# objective: for phi, compute Gini of transformed e (e**phi) and return abs difference to target
# -------------------------
def gini_diff_for_phi(phi):
    e_phi = e ** phi
    lifetimes = lifetime_incomes_from_e(e_phi, age_wgts)
    g = weighted_gini(lifetimes, abil_wgts)
    return abs(g - target_gini)

# -------------------------
# find phi via scalar minimization
# -------------------------
res = minimize_scalar(gini_diff_for_phi, bounds=(0.2, 5.0), method="bounded",
                      options={"xatol": 1e-4})
phi_hat = res.x
print("phi_hat:", phi_hat)
print("Gini at phi_hat:", weighted_gini(lifetime_incomes_from_e(e**phi_hat, age_wgts), abil_wgts))
print("Target Gini:", target_gini)

# -------------------------
# build transformed e matrix and plot
# -------------------------
e_trans = e ** phi_hat

ages = np.arange(20, 20 + S)

out_dir = "C:/Users/David Munson/Desktop/git/CompEcon_Fall25/Problemsets/ProblemSet9"
os.makedirs(out_dir, exist_ok=True)

plt.figure(figsize=(10,6))
for j in range(J):
    plt.plot(ages, e[:, j], label=f"US Group {j+1}", linestyle="-", alpha=0.6)
    plt.plot(ages, e_trans[:, j], label=f"Country Group {j+1}", linestyle="--", alpha=0.9)
plt.xlabel("Age")
plt.ylabel("Effective labor units (e)")
plt.title(f"Earnings profiles: US (solid) vs transformed country (dashed); phi={phi_hat:.3f}")
plt.legend(bbox_to_anchor=(1.02, 1), loc="upper left", ncol=1, fontsize="small")
plt.grid(alpha=0.25)
plt.tight_layout()
figpath = os.path.join(out_dir, "e_profiles_us_vs_transformed.png")
plt.savefig(figpath, dpi=150, bbox_inches="tight")
plt.show()
print("Saved figure to:", figpath)

# -------------------------
# Short numeric summary
# -------------------------
g_us = weighted_gini(lifetime_incomes_from_e(e, age_wgts), abil_wgts)
g_new = weighted_gini(lifetime_incomes_from_e(e_trans, age_wgts), abil_wgts)
print(f"US lifetime-income Gini (baseline): {g_us:.3f}")
print(f"Transformed-country lifetime-income Gini: {g_new:.3f}")
```
